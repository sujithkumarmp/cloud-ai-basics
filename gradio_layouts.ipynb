Blocks > Tabs > Rows > Columns

```python
with gr.Blocks() as demo:
    with gr.Tab("Tab One"):
        with gr.Row():
            text1 = gr.Text(value="Output One")
            text2 = gr.Text(value="Output Two")
    with gr.Tab("Tab Two"):
      with gr.Row():
        text1 = gr.Text(value="Output One")
        text2 = gr.Text(value="Output Two")
      with gr.Row():
        with gr.Column():
          text1 = gr.Text(value="Output One")
          text2 = gr.Text(value="Output Two")
        with gr.Column():
          text1 = gr.Text(value="Output three")


demo.launch()
```

Component Interaction

```python
def multiply(x,y):
    return x*y
```

```python
with gr.Blocks() as app:
    with gr.Row():
        x_slider = gr.Slider()
        y_slider = gr.Slider()

    with gr.Row():
        result= gr.Text()

    x_slider.change(fn=multiply, inputs=[x_slider, y_slider], outputs=[result])

    y_slider.change(fn=multiply, inputs=[x_slider, y_slider], outputs=[result])
app.launch()
#
```

GUI Example

```python
from transformers import AutoImageProcessor, AutoModelForImageClassification

processor = AutoImageProcessor.from_pretrained("microsoft/resnet-18")
model = AutoModelForImageClassification.from_pretrained("microsoft/resnet-18")
```

```python
def classify_image(image):
  image = processor(image, return_tensors="pt")["pixel_values"]
  logits = model(image).logits

  predicted_label = logits.argmax(-1).item()
  return model.config.id2label[predicted_label]
#
```

```python
gIface = gr.Interface(fn=classify_image,
                      inputs=gr.Image(),
                      outputs=gr.Label(),
                      title="RESR",
                      description="Upload")
gIface.launch()
```

Pipelines Integration

```python
from transformers import pipeline
sentiment_analysis = pipeline("sentiment-analysis")
```

```python
sentiment_analysis("I am sad")
```

```python
def predict_sentiment(text):
  result = sentiment_analysis(text)
  return result[0]["label"], result[0]["score"]
```

```python
predict_sentiment("I am Happy")
#
```

```python
iface = gr.Interface(fn=predict_sentiment,
                     inputs=gr.Textbox(lines=2,placeholder="Enter your review here..."),
                     outputs=[gr.Text(label="Sentiment"), gr.Number(label="Confidence Score")]  )
iface.launch()
```
